# Doubly linked list

```python
class Node :
    def __init__(self,value):
        self.value= value
        self.next = None
        self.prev = None

class Doublylinkedlist :
    def __init__(self):
        self.head = None
        self.tail = None  # useful for O(1) appends at the end
        self.n = 0

    def __len__(self):
        return  self.n
```
