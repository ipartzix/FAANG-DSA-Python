# Create a Linked list from scratch

## LinkedList class — core methods, in this order:

- __init__ — set up head = None, and track length = 0
- insert_at_head(val) — new node becomes the new head
- insert_at_tail(val) — new node goes at the end
- traverse() / __str__ — walk from head to end, print/collect values
- length() or __len__ — return node count
- insert_at_middle(index, val) — insert at arbitrary position
- delete_head() — remove first node
- delete_tail() — remove last node
- delete_by_value(val) — find and remove a node matching a value
- search(val) — return True/False or index if found

```python
class Node:
    def __init__(self, val):
        self.val = val
        self.next = None
class LinkedList:
    def __init__(self):
        self.head = None
        self.n = 0

    def __len__(self):
        return  self.n

    def insert_at_head(self, val):
        new_node = Node(val)
        new_node.next = self.head
        self.head = new_node
        self.n += 1

    def insert_at_end(self, val):
        new_node = Node(val)
        if not self.head:
            self.head = new_node
            self.n += 1
            return

        current = self.head
        while current.next:
            current = current.next
        current.next = new_node
        self.n += 1

    def traverse(self):
        cn= self.head
        while cn:
            print(cn.val)
            cn = cn.next

    def delete_head(self):
        if not  self.head:
            return None
        tmp = self.head
        self.head = self.head.next
        tmp.next = None
        self.n -= 1
        return tmp.val

    def delete_tail(self):
        if not  self.head:
            return None
        if self.head.next is None:
            val = self.head.val
            self.head = None
            self.n -=1
            return val
        current = self.head
        while current.next.next is not None:
            current = current.next
        val = current.next.val           # whose value are we deleting?
        current.next = None  # detach the tail
        self.n -= 1
        return val
```
