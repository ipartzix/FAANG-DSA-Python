```python
class Node:
    def __init__(self, val):
        self.val = val
        self.next = None
class LinkedList:
    def __init__(self):
        self.head = None
        self.n = 0

    def __len__(self):
        return  self.n

    def insert_at_head(self, val):
        new_node = Node(val)
        new_node.next = self.head
```
