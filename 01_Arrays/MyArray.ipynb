# D Array

## Methods:

- append(val) — add to the end; if storage is full (length == capacity), double the capacity and copy all existing elements into the new storage first
- pop() — remove and return the last element, shrink length
- insert(index, val) — insert at a given index, shifting everything after it one step right (resize first if full)
- delete(index) — remove element at index, shifting everything after it one step left
- search(val) — linear scan, return index if found else -1
- get(index) — return element at index (with bounds checking)
- __len__(self) — return current length
- __str__(self) — return a readable string of the elements currently in use (not the whole raw storage, just up to length)

```python
# class MyArray:
#     def __init__(self):
#         self.capacity = 1          # how much space is allocated
#         self.length = 0            # how many elements are actually stored
#         self.arr = [None] * self.capacity   # fixed-size raw storage
```

```python
class MyArray:
    def __init__(self):
        self.capacity = 1
        self.length = 0
        self.arr = [None] * self.capacity

    def append(self, val):
        if self.length == self.capacity:
            self.new_capacity = 2 * self.capacity
            new_arr = [None] * self.new_capacity
            for i in range(self.length):
                new_arr[i] = self.arr[i]
            self.arr = new_arr
            self.capacity = self.new_capacity
        self.arr[self.length] = val
        self.length += 1

    def pop(self):
        if self.length == 0:
            return None
        last_val = self.arr[self.length - 1]
        self.length -= 1
        return last_val

    def insert(self, index, val):
        if self.capacity == self.length:
            self.new_capacity = 2 * self.capacity
            new_arr = [None] * self.new_capacity
            for i in range(self.length):
                new_arr[i] = self.arr[i]
            self.arr = new_arr
            self.capacity = self.new_capacity
        for i in range(self.length, index, -1):
            self.arr[i] = self.arr[i - 1]
        self.arr[index] = val
        self.length += 1

    def delete(self, index):
        if self.length == 0:
            return None
        for i in range(index, self.length - 1):
            self.arr[i] = self.arr[i + 1]
        self.length = self.length - 1

    def search(self, val):
        for i in range(self.length):
            if self.arr[i] == val:
                 return i
        return -1

    def get(self, index):
        if 0 > index or index >= self.length:
            return None
        return self.arr[index]

    def __len__(self):
        return self.length

    def __str__(self):
        result = []
        for i in range(self.length):
            result.append(self.arr[i])
        return str(result)
```

```python
a = MyArray()
a.append(10)
a.append(20)
a.append(30)
print(a)              # [10, 20, 30]
print(len(a))         # 3
a.insert(1, 99)
print(a)              # [10, 99, 20, 30]
a.delete(1)
print(a)              # [10, 20, 30]
print(a.search(20))   # 1
print(a.search(100))  # -1
print(a.get(0))       # 10
print(a.pop())        # 30
print(a)              # [10, 20]
```
